```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageEnhance
import cv2
import warnings
warnings.filterwarnings('ignore')
```

```python
img_blurred = np.array(Image.open("blurredimage.jpeg").convert("RGB").resize((512, 512)))
img_overexposed = np.array(Image.open("overexposedimage.jpeg").convert("RGB").resize((512, 512)))
img_dark = np.array(Image.open("darkimage.jpg").convert("RGB").resize((512, 512)))
img_lowcontrast = np.array(Image.open("lowcontrastimage.jpg").convert("RGB").resize((512, 512)))
```

```python
def enhance_blurred(img):
    pil = Image.fromarray(img)
    sharpener = ImageEnhance.Sharpness(pil)
    result = sharpener.enhance(13.0)
    return np.array(result)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
fig.suptitle("Blurred Image Enhancement", fontsize=14, fontweight='bold')

axes[0].imshow(img_blurred)
axes[0].set_title("Original — Blurred")
axes[0].axis('off')

axes[1].imshow(enhance_blurred(img_blurred))
axes[1].set_title("Enhanced — Sharpened")
axes[1].axis('off')

plt.tight_layout()
plt.savefig("enhanced_blurred.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
def enhance_overexposed(img):
    pil = Image.fromarray(img)
    brightness = ImageEnhance.Brightness(pil)
    result = brightness.enhance(0.6)
    contrast = ImageEnhance.Contrast(result)
    result = contrast.enhance(1.5)
    return np.array(result)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
fig.suptitle("Overexposed Image Enhancement", fontsize=14, fontweight='bold')

axes[0].imshow(img_overexposed)
axes[0].set_title("Original — Overexposed")
axes[0].axis('off')

axes[1].imshow(enhance_overexposed(img_overexposed))
axes[1].set_title("Enhanced — Brightness & Contrast")
axes[1].axis('off')

plt.tight_layout()
plt.savefig("enhanced_overexposed.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
def enhance_dark(img):
    pil = Image.fromarray(img)
    brightness = ImageEnhance.Brightness(pil)
    result = brightness.enhance(6.0)
    return np.array(result)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
fig.suptitle("Dark Image Enhancement", fontsize=14, fontweight='bold')

axes[0].imshow(img_dark)
axes[0].set_title("Original — Dark")
axes[0].axis('off')

axes[1].imshow(enhance_dark(img_dark))
axes[1].set_title("Enhanced — Brightness Boost")
axes[1].axis('off')

plt.tight_layout()
plt.savefig("enhanced_dark.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
def enhance_lowcontrast(img):
    pil = Image.fromarray(img)
    contrast = ImageEnhance.Contrast(pil)
    result = contrast.enhance(2.0)
    return np.array(result)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
fig.suptitle("Low Contrast Image Enhancement", fontsize=14, fontweight='bold')

axes[0].imshow(img_lowcontrast)
axes[0].set_title("Original — Low Contrast")
axes[0].axis('off')

axes[1].imshow(enhance_lowcontrast(img_lowcontrast))
axes[1].set_title("Enhanced — Contrast Boost")
axes[1].axis('off')

plt.tight_layout()
plt.savefig("enhanced_lowcontrast.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
original_all = [img_blurred, img_overexposed, img_dark, img_lowcontrast]
enhanced_all = [
    enhance_blurred(img_blurred),
    enhance_overexposed(img_overexposed),
    enhance_dark(img_dark),
    enhance_lowcontrast(img_lowcontrast),
]
titles = ["Blurred", "Overexposed", "Dark", "Low Contrast"]

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
fig.suptitle("Image Enhancement — Final Summary", fontsize=14, fontweight='bold')

for col in range(4):
    axes[0, col].imshow(original_all[col])
    axes[0, col].set_title(f"Original\n{titles[col]}", fontsize=9)
    axes[0, col].axis('off')

    axes[1, col].imshow(enhanced_all[col])
    axes[1, col].set_title(f"Enhanced\n{titles[col]}", fontsize=9)
    axes[1, col].axis('off')

plt.tight_layout()
plt.savefig("final_enhanced.png", dpi=150, bbox_inches='tight')
plt.show()
```
